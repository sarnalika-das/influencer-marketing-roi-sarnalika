```python
import pandas as pd
import numpy as np

# Display all columns
pd.set_option('display.max_columns', None)

# Display wider tables
pd.set_option('display.width', None)
```

```python
influencers = pd.read_excel("influencers.xlsx")

campaigns = pd.read_excel("campaigns.xlsx")

campaign_perf = pd.read_excel("influencer campaign performance.xlsx")

orders = pd.read_excel("orders.xlsx")

products = pd.read_excel("products.xlsx")

customers = pd.read_excel("customer demographics.xlsx")
```

```python
datasets = {
    "Influencers": influencers,
    "Campaigns": campaigns,
    "Campaign Performance": campaign_perf,
    "Orders": orders,
    "Products": products,
    "Customers": customers
}
```

```python
for name, df in datasets.items():
    print("="*70)
    print(f"{name}")
    print("="*70)

    print("\nShape:")
    print(df.shape)

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nFirst 5 rows:")
    display(df.head())

    print("\n")
```

```python
# ============================================
# Missing Values Summary
# ============================================

for name, df in datasets.items():

    print("="*70)
    print(name)
    print("="*70)

    missing = df.isnull().sum()

    missing = missing[missing > 0]

    if missing.empty:
        print("✅ No missing values found.\n")

    else:
        missing_df = (
            pd.DataFrame({
                "Missing Count": missing,
                "Missing %": (missing / len(df) * 100).round(2)
            })
            .sort_values(by="Missing Count", ascending=False)
        )

        display(missing_df)

    print()
```

```python
# ============================================
# Duplicate Record Assessment
# ============================================

for name, df in datasets.items():

    print("=" * 70)
    print(name)
    print("=" * 70)

    duplicate_rows = df.duplicated().sum()

    print(f"Duplicate Rows : {duplicate_rows}")

    print()
```

```python
# ============================================
# Primary Key Uniqueness Check
# ============================================

primary_keys = {
    "Influencers": "Influencer_ID",
    "Campaigns": "Campaign_ID",
    "Products": "Product_ID",
    "Customers": "Customer_ID",
    "Orders": "Order_ID"
}

for dataset_name, key in primary_keys.items():

    df = datasets[dataset_name]

    duplicates = df[key].duplicated().sum()

    print("="*70)
    print(dataset_name)
    print("="*70)

    print(f"Primary Key : {key}")
    print(f"Duplicate Keys : {duplicates}")

    if duplicates == 0:
        print("✅ Primary key is unique.\n")
    else:
        print("❌ Duplicate primary keys found.\n")
```

```python
# ============================================
# Composite Key Check
# ============================================

duplicate_composite = campaign_perf.duplicated(
    subset=["Source_Influencer_ID", "Campaign_ID"]
).sum()

print("Duplicate Composite Keys:", duplicate_composite)

if duplicate_composite == 0:
    print("✅ Composite key is unique.")
else:
    print("❌ Duplicate composite keys found.")
```

```python
# ============================================
# Referential Integrity Check
# ============================================

checks = [
    ("Orders", orders["Customer_ID"], customers["Customer_ID"], "Customer_ID"),
    ("Orders", orders["Product_ID"], products["Product_ID"], "Product_ID"),
    ("Orders", orders["Campaign_ID"], campaigns["Campaign_ID"], "Campaign_ID"),
    ("Orders", orders["Source_Influencer_ID"], influencers["Influencer_ID"], "Source_Influencer_ID"),

    ("Campaign Performance", campaign_perf["Campaign_ID"], campaigns["Campaign_ID"], "Campaign_ID"),
    ("Campaign Performance", campaign_perf["Source_Influencer_ID"], influencers["Influencer_ID"], "Source_Influencer_ID"),
    (
        "Campaign Performance",
        campaign_perf["Main_Product_ID"].dropna(),
        products["Product_ID"],
        "Main_Product_ID"
    )
]

for table, child, parent, key in checks:

    invalid = (~child.isin(parent)).sum()

    print("=" * 70)
    print(table)
    print("=" * 70)

    print(f"Foreign Key : {key}")
    print(f"Invalid References : {invalid}")

    if invalid == 0:
        print("✅ Referential integrity maintained.\n")
    else:
        print("❌ Invalid foreign key references found.\n")
```

```python
# ============================================
# Categorical Consistency Check
# ============================================

categorical_columns = {
    "Influencers": ["Gender", "Platform", "Tier", "Location"],
    "Campaigns": ["Product_Category", "Goal_Type", "Target_Gender", "Product_Focus_Type", "Platform"],
    "Campaign Performance": ["Product_Focus_Type"],
    "Orders": ["Channel"],
    "Products": ["Category", "Sub_Category", "Gender_Target", "Season"],
    "Customers": ["Gender", "Location", "Income_Bracket", "Preferred_Style"]
}

for dataset_name, columns in categorical_columns.items():

    print("="*80)
    print(dataset_name)
    print("="*80)

    df = datasets[dataset_name]

    for col in columns:
        print(f"\n{col}:")
        print(sorted(df[col].dropna().unique()))

    print("\n")
```

```python
# ============================================
# Outlier Detection using IQR
# ============================================

numerical_columns = {
    "Influencers": ["Follower_Count", "Avg_Cost_Per_Post", "Audience_Location_Match_pct"],
    "Campaigns": ["Budget", "Conversion_Rate_Target_pct"],
    "Campaign Performance": ["Conversions", "Revenue_Generated", "Cost", "ROI"],
    "Orders": ["Quantity", "Unit_Price", "Discount_Amount", "Total_Amount"],
    "Products": ["Price"],
    "Customers": ["Age", "Lifetime_Value", "Avg_Order_Value"]
}

for dataset_name, cols in numerical_columns.items():

    print("="*80)
    print(dataset_name)
    print("="*80)

    df = datasets[dataset_name]

    for col in cols:

        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)

        IQR = Q3 - Q1

        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        outliers = df[(df[col] < lower) | (df[col] > upper)]

        print(f"{col}: {len(outliers)} outliers")

    print()
```
